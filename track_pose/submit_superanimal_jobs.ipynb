# SuperAnimal Slurm submission

`jobs.json` is the single source of truth for paths, cage maps, output layout, skipping, and batch sizes. Edit that file, then use this notebook only to list, preview, and explicitly submit named profiles.

The default cells are dry-run only. A job is submitted only from the final cell with `submit=True`.

```python
from pathlib import Path

from submit_superanimal_jobs import job_from_profile, load_profiles
from superanimal_submit import preview_job, submit_job, submit_many

PROJECT_ROOT = Path.cwd()  # Open this notebook from the super_animal directory.
JOBS_FILE = PROJECT_ROOT / 'jobs.json'
profiles, config_directory = load_profiles(JOBS_FILE)
```

```python
# Inspect names defined in jobs.json.
print(*sorted(profiles), sep='\n')

# The active MLb2 profile scans the experiment recursively, processes only
# videos-raw directories, writes each cage's sibling tracks directory, and
# skips videos with final adapted H5/JSON output.
```

```python
# Safe preview of one named job.
PROFILE_NAME = 'mlb2_experiment'
job = job_from_profile(PROFILE_NAME, profiles[PROFILE_NAME], config_directory)
print(preview_job(job))
```

```python
# Safe preview of several profiles. Edit the list; no job is submitted here.
PROFILE_NAMES = ['mlb2_experiment']
jobs = [job_from_profile(name, profiles[name], config_directory) for name in PROFILE_NAMES]
print(*submit_many(jobs), sep='\n')
```
